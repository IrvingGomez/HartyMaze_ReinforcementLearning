# Play the Maze in Colab -- click buttons to move Harty

Browser-friendly version of `HartyPlay.ipynb`. No popup window, no
keyboard required. Use the on-screen buttons (Up / Down / Left / Right)
to move Harty one cell at a time. Press **Reset** to start over.

Goal: reach the treasure. Avoid holes. Portals teleport you.


## Set up the maze

Same variables as in `HartyMazeSolver.ipynb`. Edit then rerun.


```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd() / 'src'))

from environment import DEFAULT_MAZE, BIG_MAZE
from play_mode import launch_game_widget

MAZE = BIG_MAZE
TREASURE = (0, 15)
HARTY = (11, 12)

HOLES = [(2,14),(14,16)]
PORTALS = [((3,0),(14,12)),((8,16),(12,17))]

rows, cols = MAZE.shape
assert MAZE.ndim == 2
assert MAZE.min() >= 0 and MAZE.max() <= 14
assert 0 <= TREASURE[0] < rows and 0 <= TREASURE[1] < cols
assert 0 <= HARTY[0] < rows and 0 <= HARTY[1] < cols
assert TREASURE != HARTY

print(f'Maze: {rows}x{cols}, treasure at {TREASURE}, Harty starts at {HARTY}')
```

## Play!

Run the cell below. The maze appears on the left, the controls on the
right. Click a direction button to move. Click **Reset** any time to
send Harty back to his starting cell.


```python
game = launch_game_widget(MAZE, TREASURE, HARTY, holes=HOLES, portals=PORTALS)
```

## Inspect the live state

The cell above returns a dict whose `'current'` key always points to
the latest `PlayState`. Run the cell below after playing to see where
Harty ended up.


```python
state = game['current']
print(f'Position: {state.position}')
print(f'Facing:   {state.facing}')
print(f'Steps:    {state.steps_taken}')
print(f'Won?      {state.won}')
print(f'Dead?     {state.dead}')
```
